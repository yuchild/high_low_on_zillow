# Bay Area Zillow EDA

## Goals
- Inspect Bay Area county home price and rent trends
- Compare latest levels by county
- Compare MoM and YoY changes
- Identify dashboard-ready KPIs

## Imports

```python
import pandas as pd
import plotly.express as px

from high_low_on_zillow.paths import PROCESSED_DIR
```

```python
home = pd.read_parquet(PROCESSED_DIR / "bay_area_home_prices.parquet")
rent = pd.read_parquet(PROCESSED_DIR / "bay_area_rentals.parquet")
home_kpis = pd.read_parquet(PROCESSED_DIR / "bay_area_home_price_kpis.parquet")
rent_kpis = pd.read_parquet(PROCESSED_DIR / "bay_area_rent_kpis.parquet")
```

```python
home.head(), rent.head()
```

```python
home_kpis.sort_values("home_price", ascending=False)
```

```python
rent_kpis.sort_values("rent", ascending=False)
```

```python
fig = px.line(
    home,
    x="date",
    y="home_price",
    color="county",
    title="Bay Area County Home Prices"
)
fig.show()
```

```python
fig = px.line(
    rent,
    x="date",
    y="rent",
    color="county",
    title="Bay Area County Rents"
)
fig.show()
```
